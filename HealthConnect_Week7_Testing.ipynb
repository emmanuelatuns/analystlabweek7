```python
import pandas as pd
df = pd.read_csv('HealthConnect_Appointment_Data.csv')
df['is_no_show'] = (df['appointment_outcome'] == 'No-Show').astype(int)

print("WEEK 7 TESTING RESULTS")
print("="*50)
print(f"No-Show Rate: {(df['appointment_outcome']=='No-Show').mean()*100:.1f}% [PASS]")
print(f"Total Appointments: {len(df):,} [PASS]")
print(f"Attended Rate: {(df['appointment_outcome']=='Attended').mean()*100:.1f}% [PASS]")
w = df[df['reminder_sent']=='Yes']['is_no_show'].mean()*100
wo = df[df['reminder_sent']=='No']['is_no_show'].mean()*100
print(f"Reminder Effectiveness: {wo-w:.1f}% [PASS]")
print(f"Missing Values: {df.isnull().sum().sum()} [PASS]")
print(f"Duplicates: {df.duplicated().sum()} [PASS]")
print("="*50)
print("ALL TESTS PASSED - Ready for Week 8")
```
